## Question 1

AiiDAlab offers different types of calculations for adsorption studies:

- **Gas adsorption isotherms:** Grand Canonical Monte Carlo (GCMC) simulations are used to calculate the gas uptake as a function of pressure at a given temperature.

- **Henry coefficients:** these characterize adsorption in the low-pressure limit and can be used to estimate the affinity and selectivity of a material toward different gases.

- **Pore analysis:** structural properties such as the accessible pore volume, accessible surface area and porosity are calculated to characterize the porous material.

- **Density calculations:** the framework density is calculated and provides an additional structural descriptor of the porous material.

These tools can be applied both to porous frameworks and to surfaces like LSMO, supporting adsorption studies across materials classes.

$$
K_H(CH_4) = 1.12046 \cdot 10^{-6} \ \text{mol} \cdot \text{kg}^{-1} \cdot \text{Pa}^{-1}
$$

$$
K_H(CO_2) = 4.95454 \cdot 10^{-6} \ \text{mol} \cdot \text{kg}^{-1} \cdot \text{Pa}^{-1}
$$
At 300K, $CO_2$ Henry coefficient is almost four times higher than $CH_4$ Henry coefficient, wich means that the $CO_2$ molecules interact more stongly with IRMOF-1 framework than $CH_4$ molecule.

### Question 5


The values computed were once again obtained using AiiDAlab. 
The code used to generate the graph is shown below:

```python
import pyiast
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df_ch4 = pd.read_csv("CH4_IRMOF1.csv")
df_co2 = pd.read_csv("CO2_IRMOF1.csv")

co2_isotherm = pyiast.InterpolatorIsotherm(df_co2,
    loading_key="Loading(mmol/g)",
    pressure_key="Pressure(bar)"
)

ch4_isotherm = pyiast.InterpolatorIsotherm(
    df_ch4,
    loading_key="Loading(mmol/g)",
    pressure_key="Pressure(bar)"
)
```

```python
y=np.array([0.6, 0.4])

q_ch4=[]
q_co2=[]

pressures = np.arange(0.1, 4.01, 0.2)

for P in pressures:
    q = pyiast.iast(P * y, [ch4_isotherm, co2_isotherm])
    q_ch4.append(q[0])
    q_co2.append(q[1])
```

```python
plt.plot(pressures, q_co2, "o-", label="CO2")
plt.plot(pressures, q_ch4, "o-", label="CH4")
plt.xlabel("Total Pressure (bar)")
plt.ylabel("Loading (mmol/g)")
plt.legend()
plt.show()
```
