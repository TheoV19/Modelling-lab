### Question 5


```python
import pyiast
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df_ch4 = pd.read_csv("CH4_IRMOF1.csv")
df_co2 = pd.read_csv("CO2_IRMOF1.csv")

co2_isotherm = pyiast.InterpolatorIsotherm(df_co2,
    loading_key="Loading(mmol/g)",
    pressure_key="Pressure(bar)"
)

ch4_isotherm = pyiast.InterpolatorIsotherm(
    df_ch4,
    loading_key="Loading(mmol/g)",
    pressure_key="Pressure(bar)"
)
```

```python
y=np.array([0.6, 0.4])

q_ch4=[]
q_co2=[]

pressures = np.arange(0.1, 4.01, 0.2)

for P in pressures:
    q = pyiast.iast(P * y, [ch4_isotherm, co2_isotherm])
    q_ch4.append(q[0])
    q_co2.append(q[1])
```

```python
plt.plot(pressures, q_co2, "o-", label="CO2")
plt.plot(pressures, q_ch4, "o-", label="CH4")
plt.xlabel("Total Pressure (bar)")
plt.ylabel("Loading (mmol/g)")
plt.legend()
plt.show()
```

### Question 6

```python
total_pressure= np.array([0.1, 1, 2, 3])
y = np.array([0.6, 0.4]) 
for P in total_pressure:
    q = pyiast.iast(P * y, [ch4_isotherm, co2_isotherm])
    q_ch4=q[0]
    q_co2=q[1]
    S= (q_co2 / q_ch4) * (0.6 / 0.4)
    print(P,S)
```

It can be deduced that the selectivity ratio does not vary with the modification of the total pressure. This indicates that CO2 is absorbed more in general compared to CH4.
