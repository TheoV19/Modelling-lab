## Question 1

AiiDAlab offers different types of calculations for adsorption studies:

- **Gas adsorption isotherms:** Grand Canonical Monte Carlo (GCMC) simulations are used to calculate the gas uptake as a function of pressure at a given temperature.

- **Henry coefficients:** these characterize adsorption in the low-pressure limit and can be used to estimate the affinity and selectivity of a material toward different gases.

- **Pore analysis:** structural properties such as the accessible pore volume, accessible surface area and porosity are calculated to characterize the porous material.

- **Density calculations:** the framework density is calculated and provides an additional structural descriptor of the porous material.

These tools can be applied both to porous frameworks and to surfaces like LSMO, supporting adsorption studies across materials classes.

### Question 5


The values computed were once again obtained using AiiDAlab. 
The code used to generate the graph is shown below:

```python
import pyiast
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df_ch4 = pd.read_csv("isotherm_CH4.csv")
df_co2 = pd.read_csv("isotherm_CO2.csv")

co2_isotherm = pyiast.InterpolatorIsotherm(df_co2,
    loading_key="loading_absolute_average",
    pressure_key="pressure"
)

ch4_isotherm = pyiast.InterpolatorIsotherm(
    df_ch4,
    loading_key="loading_absolute_average",
    pressure_key="pressure"
)
```

```python
y=np.array([0.6, 0.4])

q_ch4=[]
q_co2=[]

pressures = np.arange(0.1, 4.01, 0.2)

for P in pressures:
    q = pyiast.iast(P * y, [ch4_isotherm, co2_isotherm])
    q_ch4.append(q[0])
    q_co2.append(q[1])
```

```python
plt.plot(pressures, q_co2, "o-", label="CO2")
plt.plot(pressures, q_ch4, "o-", label="CH4")
plt.xlabel("Total Pressure (bar)")
plt.ylabel("Loading (mmol/g)")
plt.legend()
plt.show()
```

## Question 6

```python
total_pressure= np.array([0.1, 1, 2, 3])
y = np.array([0.6, 0.4]) 
for P in total_pressure:
    q = pyiast.iast(P * y, [ch4_isotherm, co2_isotherm])
    q_ch4=q[0]
    q_co2=q[1]
    S= (q_co2 / q_ch4) * (0.6 / 0.4)
    print(P,S)
```

All selectivity values shows that IRMOF-1 has prefernece for CO2 over CH4. The selectivity value seems to increase slighly from low pressure to high pressure, but stays generally constant over high pressure. therefore it can be said that the selectivity has little dependence on pressure.

# Question 7


```python
total_pressure = np.array([0.1, 1, 2, 3])

y_original = np.array([0.6, 0.4])

y_new = np.array([0.8, 0.2])

S_original = []
S_new = []

for P in total_pressure:

    q = pyiast.iast(P * y_original, [ch4_isotherm, co2_isotherm])

    q_ch4 = q[0]
    q_co2 = q[1]

    S1 = (q_co2 / q_ch4) * (0.6 / 0.4)
    S_original.append(S1)

    q = pyiast.iast(P * y_new, [ch4_isotherm, co2_isotherm])

    q_ch4 = q[0]
    q_co2 = q[1]

    S2 = (q_co2 / q_ch4) * (0.8 / 0.2)
    S_new.append(S2)

    print(P, S1, S2)

```

```python
plt.plot(total_pressure, S_original, "o-", label="yCO2 = 0.4")
plt.plot(total_pressure, S_new, "o-", label="yCO2 = 0.2")

plt.xlabel("Pressure (bar)")
plt.ylabel("CO2/CH4 selectivity")
plt.legend()
plt.show()
```

```python
for P, S1, S2 in zip(total_pressure, S_original, S_new):

    difference = abs(S2 - S1) / S1 * 100

    print(P, difference, "%")
```

The IAST calculations were repeated by decreasing the CO2 molar fraction from 0.4 to 0.2, while increasing the CH4 molar fraction from 0.6 to 0.8. The calculated CO2/CH4 selectivities remain approximately unchanged, showing that the selectivity of IRMOF-1 is only weakly dependent on the gas-phase composition under these conditions.

To determine whether selectivity depends on temperature, pure CO2 and CH4 adsorption isotherms should be calculated at several temperatures. IAST calculations can then be repeated at each temperature while keeping the pressure and gas composition constant. Comparing the resulting selectivities would reveal the temperature dependence.

```python
q_max = df_ch4["loading_absolute_average"].iloc[-1]
P_max = df_ch4["pressure"].iloc[-1]

M_CH4 = 16.04  # g/mol

mass_CH4 = q_max * M_CH4
wt_percent = mass_CH4 / 1000 * 100

print("Maximum pressure:", P_max, "bar")
print("CH4 loading:", q_max, "mol/kg")
print("CH4 mass stored:", mass_CH4, "g/kg")
print("Gravimetric uptake:", wt_percent, "%")
```

IRMOF-1 appears to be a promising material for CH4 storage. It has a high porosity of approximately 77.5%, providing a large accessible volume for gas adsorption. The CH4 adsorption isotherm continuously increases with pressure and reaches approximately 1.13 Kg/mol at 10.2 bar. No clear saturation is observed within the investigated pressure range, suggesting that a larger amount of CH4 could be stored at higher pressures. Therefore, the high porosity and increasing methane uptake indicate that IRMOF-1 is a promising candidate for CH4 storage.
