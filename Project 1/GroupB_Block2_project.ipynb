**The pure-component isotherms for $\mathrm{CO}_2$ and $\mathrm{N}_2$ at $25^\circ\mathrm{C}$ are plotted for each given structure.**

### MG-MOF-74

```python
N2_mg = pd.read_csv('N2_MG.csv')

plt.figure(figsize=(8, 5))

x1=N2_mg["pressure"]
y1=N2_mg["loading_absolute_average"]

plt.plot(x1, y1, "o-", label="N2 isotherm", color="red", markersize=6)

plt.xlabel('Pressure (bar)', fontsize=12)
plt.ylabel('Loading (mmol/kg)', fontsize=12)
plt.title('CO₂ and N₂ Pure Isotherms at 298K, MG-MOF-74', fontsize=14, fontweight='bold')
plt.legend(fontsize=11, loc='best')
plt.legend()
plt.show()
```
