### a. Compute missing Langmuir parameters

#### After the researcher did the fitting, she obtained the following values:
    qsat = 4.72 mmol/g
    k(263) = 6.75e-06 1/Pa
    k(303) = 1.48e-06 1/Pa
    k(343) = 5.16e-07 1/Pa

As we know the temperature and their corresponding value of K, we can plot $\ln K$ as a function of $\frac{1}{T}$.

```python

temperatures=np.array([263,303,343])
K_values=np.array([6.75e-06, 1.48e-06, 5.16e-07])

x = 1 / temperatures
y = np.log(K_values)

slope, intercept = np.polyfit(x, y, 1)

plt.plot(x, y, "o", label="Data")
plt.plot(x, slope * x + intercept, label=f"Slope = {slope:.2f} K")
plt.xlabel("1/T (K⁻¹)")
plt.ylabel("ln K")
plt.legend()
plt.show()

print(f"Slope: {slope:.2f} K")
print(f"Intercept: {intercept:.2f}")
```

From exercise 4, we know that

$$
\ln k(T) = \ln(k_0) - \frac{\Delta H}{R}\cdot\frac{1}{T}.
$$

Therefore, $\ln(k_0)$ is the intercept with the y-axis and $\Delta H$ = -R $\cdot$ slope. qsat was already given and is equal to **qsat = 4.72 mmol/g**.

```python
R = 8.314 #J/(molK)

qsat = 4.72 #mmol/g

k_0 = np.exp(intercept) #1/Pa

delta_H = (-R * slope)/1000 #kJ/mol

print("qsat = 4.72 mmol/g")
print(f"k0 = {k_0:.3e} Pa⁻¹")
print(f"ΔH = {delta_H:.2f} kJ/mol")

```

A shown in question 2.b, Henry coefficients for each temperature are defined as:
$$
K_H(T) = q_{sat} \cdot k(T),
$$


```python
henry_coeff = qsat * K_values

for T, K_H in zip(temperatures, henry_coeff):
    print(f"T={T}, K_H = {K_H: .3e} mol/(kg·Pa)")
```

Henry’s coefficient decreases as temperature increases. Therefore, at the same low pressure, less CO₂ is adsorbed at higher temperatures. This is consistent with the negative adsorption enthalpy: adsorption is exothermic, so increasing temperature disfavors adsorption.
