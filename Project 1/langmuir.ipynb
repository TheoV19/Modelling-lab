```python
## Schematic Graph for Langmuir
#type:ignore
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 10, 100)
y = 2 * x + 1

plt.plot(x, y, color="black")
plt.xlim(left=0)
plt.xlabel("P")
plt.ylabel("1/q")
plt.title("Schematic graph of the Equation derived from the Langmuir Equation")

# Hide numerical ticks for a schematic appearance
plt.xticks([])
plt.yticks([])

plt.savefig("schematic_graph_Langmuir.pdf", bbox_inches="tight")
plt.show()
```

```python
## Schematic Graph for Van`t Hoff
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 10, 100)
y = -2 * x + 1

plt.plot(x, y, color="black")
plt.xlim(left=0)
plt.xlabel("1/T")
plt.ylabel("ln(k)")
plt.title("Schematic graph of the Equation derived from the Vant't Hoff Equation")

# Hide numerical ticks for a schematic appearance
plt.xticks([])
plt.yticks([])

plt.savefig("schematic_graph_Vant_Hoff.pdf", bbox_inches="tight")
plt.show()
```
